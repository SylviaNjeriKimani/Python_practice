```python
#for loop
#counting from 1 to 10
#using a range
```

```python
for number in range(1,11):
  print(number)
```

```python
#using a list
fruits=["apple","banana","mango","orange"]
for fruit in fruits:
  print(fruit)
```

```python
#for loop with a condition
#print even numbers
for nos in range(1,11):
  if nos % 2 == 0:
    print(nos)
```
